import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# 1. Verify that your browser notebook is successfully using the Cloud GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"--- Environment Check ---")
print(f"Using device: {device.upper()}")
if torch.cuda.is_available():
    print(f"GPU Model: {torch.cuda.get_device_name(0)}\n")
else:
    print("WARNING: GPU is not active! Please change your runtime settings to GPU.\n")

# 2. Generate dummy data (Simulating a simple tabular dataset)
# 1000 samples, 20 features each
X = torch.randn(1000, 20)
# Binary labels (0 or 1)
y = torch.randint(0, 2, (1000, 1)).float()

dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_batch_size=32, shuffle=True)

# 3. Build a simple neural network model
class SimpleModel(nn.Module):
    def __init__(self):
        super(SimpleModel, self).__init__()
        self.network = nn.Sequential(
            nn.Linear(20, 64),
            nn.ReLU(),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.network(x)

# Initialize model, loss function, and optimizer
model = SimpleModel().to(device)  # Sends the model parameters straight to the Cloud GPU
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# 4. Run a quick 5-epoch training loop entirely on the Cloud GPU
print(f"--- Starting Training Test ---")
for epoch in range(5):
    epoch_loss = 0.0
    for batch_X, batch_y in dataloader:
        # Move your data chunks to the Cloud GPU memory
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)
        
        # Forward pass
        predictions = model(batch_X)
        loss = criterion(predictions, batch_y)
        
        # Backward pass (optimization)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        epoch_loss += loss.item()
        
    print(f"Epoch {epoch+1}/5 | Average Loss: {epoch_loss/len(dataloader):.4f}")

print("\nSuccess! Your online cloud GPU setup is fully configured and ready.")
